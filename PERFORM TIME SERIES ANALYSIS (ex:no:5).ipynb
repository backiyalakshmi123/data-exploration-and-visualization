```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load sample time series data
df = sns.load_dataset("flights")

# Convert to datetime
df['Date'] = pd.to_datetime(df['year'].astype(str) + '-' + df['month'].astype(str))
df.set_index('Date', inplace=True)

# Line Plot
plt.plot(df['passengers'])
plt.title("Monthly Passengers")
plt.xlabel("Year")
plt.ylabel("Passengers")
plt.show()

# Rolling Mean (12 months)
df['Rolling'] = df['passengers'].rolling(12).mean()
df[['passengers', 'Rolling']].plot()
plt.title("Rolling Mean (12 Months)")
plt.show()

# Boxplot by Month
sns.boxplot(x='month', y='passengers', data=df.reset_index())
plt.title("Monthly Seasonality")
plt.show()
```
