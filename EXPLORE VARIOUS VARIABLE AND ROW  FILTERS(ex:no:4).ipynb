```python
import pandas as pd
import matplotlib.pyplot as plt

# STEP 1: Create sample dataset
data = {
    'Name': ['A', 'B', 'C', 'D', 'E', None],
    'Age': [25, 30, 22, None, 28, 27],
    'Marks': [88, 92, None, 75, 85, 90],
    'Passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
print("🔹 Original Data:\n", df)

# STEP 2: Remove rows with missing values
df_cleaned = df.dropna()
print("\n🔹 After Removing Rows with Missing Values:\n", df_cleaned)

# STEP 3: Filter rows where Marks > 85
df_filtered = df_cleaned[df_cleaned['Marks'] > 85]
print("\n🔹 Filtered Rows (Marks > 85):\n", df_filtered)

# STEP 4: Select only 'Name' and 'Marks' columns
df_selected = df_filtered[['Name', 'Marks']]
print("\n🔹 Selected Columns:\n", df_selected)

# STEP 5: Plotting
plt.figure(figsize=(6, 4))
plt.bar(df_selected['Name'], df_selected['Marks'], color='skyblue')
plt.title('Students with Marks > 85')
plt.xlabel('Name')
plt.ylabel('Marks')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
