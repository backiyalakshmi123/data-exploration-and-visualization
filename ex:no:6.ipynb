```python
import pandas as pd
import numpy as np
import folium

# Step 1: Create random dataset
np.random.seed(42)
data = pd.DataFrame({
    'LATITUDE': np.random.uniform(37.5, 38.5, 100),
    'LONGITUDE': np.random.uniform(-123.0, -121.0, 100),
    'VALUE': np.random.randint(1, 100, 100)
})

# Step 2: Display summary (like your image)
print(data.describe())

# Step 3: Create interactive map
mymap = folium.Map(location=[data['LATITUDE'].mean(), data['LONGITUDE'].mean()], zoom_start=8)

# Add points with tooltips
for _, row in data.iterrows():
    folium.CircleMarker(
        location=[row['LATITUDE'], row['LONGITUDE']],
        radius=5,
        fill=True,
        fill_color='blue',
        color='gray',
        fill_opacity=0.7,
        popup=f"Value: {row['VALUE']}",
        tooltip=f"Lat: {row['LATITUDE']:.2f}, Lon: {row['LONGITUDE']:.2f}"
    ).add_to(mymap)

# Step 4: Save map
mymap.save("random_map.html")
print("Map saved as 'random_map.html'")
```
